### Day 2 reflections:
Resampling historical NAV values to fill weekend gaps was extremely painfull. 
I tried calling `resample('B')` on the grouped dataframe but kept getting index and datetime errors because index was not sorted or was a normal object index instead of datetime. 
Googled it and found a Stack Overflow link which saved me:
https://stackoverflow.com/questions/44123432/pandas-groupby-resample-fill-values

```python
import pandas as pd
RAW_DIR = Path('../data/raw')
df_nav = pd.read_csv(RAW_DIR / '02_nav_history.csv')
# Forgot to set date index or convert type to datetime first!
df_nav.groupby('amfi_code')['nav'].resample('B').ffill()
```

```python
# Got it to work after setting index to datetime
df_nav['date'] = pd.to_datetime(df_nav['date'])
df_nav_indexed = df_nav.set_index('date')
df_clean = df_nav_indexed.groupby('amfi_code')['nav'].resample('B').ffill().reset_index()
print(f"Successfully resampled and forward-filled gaps: {len(df_clean)} rows.")
```
